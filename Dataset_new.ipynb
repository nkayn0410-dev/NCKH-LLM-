```python
import subprocess
import re

def check_commit(commit_hash):
    """Kiểm tra 1 commit có hợp lệ không. Trả về (hợp_lệ, lý_do)."""

    # --- Đếm số parent ---
    r1 = subprocess.run(
        ["git", "show", "--no-patch", "--format=%P", commit_hash],
        capture_output=True, text=True
    )
    num_parents = len(r1.stdout.strip().split())

    # Nếu là merge commit (>1 parent) → loại
    if num_parents > 1:
        return (False, "Merge commit")

    # --- Đếm số file và dòng thay đổi ---
    r2 = subprocess.run(
        ["git", "show", "--shortstat", "--format=", commit_hash],
        capture_output=True, text=True
    )
    stat = r2.stdout

    # Dùng regex tìm các con số
    files = re.search(r'(\d+) file', stat)
    insertions = re.search(r'(\d+) insertion', stat)
    deletions = re.search(r'(\d+) deletion', stat)

    num_files = int(files.group(1)) if files else 0
    num_ins = int(insertions.group(1)) if insertions else 0
    num_del = int(deletions.group(1)) if deletions else 0
    total_lines = num_ins + num_del

    # --- Áp dụng các luật lọc ---
    if num_files > 10:
        return (False, f"Quá nhiều file ({num_files})")
    if total_lines > 500:
        return (False, f"Diff quá lớn ({total_lines} dòng)")
    if total_lines < 5:
        return (False, f"Diff quá nhỏ ({total_lines} dòng)")

    # Qua hết các luật → hợp lệ
    return (True, f"OK ({num_files} file, {total_lines} dòng)")
```

```python
# Danh sách 5 hash thật của bạn
commit_hashes = [
    "fb4415d8aee6c1",   # CVE-2023-38545
    "076a2f629119222a",   # CVE-2023-23914
    "a64e3e59938abd7d6",   # CVE-2022-32221
    "139a54ed0a172ada",   # CVE-2022-27774
    "43157490a5054bd"
]

# Về đúng thư mục curl
os.chdir('/content/drive/MyDrive/NCKH_dataset/curl')

# Lọc từng commit
print("=== KẾT QUẢ LỌC COMMIT ===\n")
valid_commits = []   # danh sách chứa commit hợp lệ

for h in commit_hashes:
    is_valid, reason = check_commit(h)
    status = "✅ GIỮ" if is_valid else "❌ LOẠI"
    print(f"{status} | {h} | {reason}")

    if is_valid:
        valid_commits.append(h)

print(f"\n📊 Kết quả: giữ {len(valid_commits)}/{len(commit_hashes)} commit")
print("Commit hợp lệ:", valid_commits)
```

```python
# --- GÁN NHÃN CHO 1 DIFF ---
current_hash = "fb4415d8aee6c1"
# Tạo dictionary lưu nhãn (rỗng lúc đầu)
labels = {}
# Thông tin bạn quyết định (CHỈNH SỬA 3 CÁI NÀY):
bug_description = "SOCKS5 heap buffer overflow: khi hostname dài > 255 ký tự, curl cố chuyển sang local resolve thay vì báo lỗi, dẫn đến tràn bộ đệm heap (CVE-2023-38545)"
vulnerable_code = "socks5_resolve_local = TRUE;"
fixed_code = "return CURLPX_LONG_HOSTNAME;"
reason = "Code cũ xử lý hostname quá dài bằng cách bật cờ socks5_resolve_local=TRUE để chuyển sang local resolve, nhưng logic này bị sai và gây heap overflow."

# --- Ghi vào labels ---
labels[current_hash] = {
    "bug_description": bug_description,
    "vulnerable_code": vulnerable_code,
    "fixed_code": fixed_code,
    "reason": reason,
    "label_vulnerable": 1,  # commit này SỬA lỗi, nên code CŨ có lỗi
    "label_fixed": 0,
}

print(f"✅ Đã lưu nhãn cho {current_hash}")
print(labels[current_hash])
```

```python
import pandas as pd

# Chuyển dict thành DataFrame
# Mỗi key của dict → 1 cột
df = pd.DataFrame.from_dict(labels, orient='index')

# Reset index để commit_hash thành 1 cột bình thường
df.reset_index(inplace=True)
df.rename(columns={'index': 'commit_hash'}, inplace=True)

# Xem 3 dòng đầu
print("📊 Dataset preview:\n")
print(df.head())
```

```python
import subprocess
import re
import os

# Về thư mục curl repo
os.chdir('/content/drive/MyDrive/NCKH_dataset/curl')

# Tạo các cột metadata
metadata = []

for h in df['commit_hash']:
    # Lấy ngày commit
    r_date = subprocess.run(
        ["git", "show", "-s", "--format=%ci", h],
        capture_output=True, text=True
    )
    commit_date = r_date.stdout.strip()

    # Lấy tác giả
    r_author = subprocess.run(
        ["git", "show", "-s", "--format=%an", h],
        capture_output=True, text=True
    )
    author = r_author.stdout.strip()

    # Lấy số file và số dòng thay đổi
    r_stat = subprocess.run(
        ["git", "show", "--shortstat", "--format=", h],
        capture_output=True, text=True
    )
    stat = r_stat.stdout

    files = re.search(r'(\d+) file', stat)
    insertions = re.search(r'(\d+) insertion', stat)
    deletions = re.search(r'(\d+) deletion', stat)

    num_files = int(files.group(1)) if files else 0
    num_ins = int(insertions.group(1)) if insertions else 0
    num_del = int(deletions.group(1)) if deletions else 0

    metadata.append({
        'commit_date': commit_date,
        'author': author,
        'num_files_changed': num_files,
        'lines_added': num_ins,
        'lines_deleted': num_del,
        'total_lines_changed': num_ins + num_del
    })

# Chuyển metadata thành DataFrame
df_meta = pd.DataFrame(metadata)

# Gộp với df chính
df = pd.concat([df, df_meta], axis=1)

print("✅ Đã thêm metadata. Xem lại:\n")
print(df.head())
```

```python
# Chọn các cột quan trọng theo thứ tự
columns_to_keep = [
    'commit_hash',
    'commit_date',
    'author',
    'bug_description',
    'vulnerable_code',
    'fixed_code',
    'reason',
    'label_vulnerable',
    'label_fixed',
    'num_files_changed',
    'lines_added',
    'lines_deleted',
    'total_lines_changed'
]

df_final = df[columns_to_keep]

print("📊 Dataset cuối cùng:\n")
print(df_final)
print(f"\n📈 Kích thước: {len(df_final)} commits, {len(df_final.columns)} cột")
```

```python
# Đường dẫn lưu file
output_path = '/content/drive/MyDrive/NCKH_dataset/curl_vulnerability_dataset.csv'

# Lưu CSV
df_final.to_csv(output_path, index=False, encoding='utf-8')

print(f"✅ Đã lưu dataset: {output_path}")
print(f"📦 Kích thước file: {os.path.getsize(output_path)} bytes")
```
